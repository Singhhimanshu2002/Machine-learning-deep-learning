### **2. NumPy Essentials: Array Operations & Vectorization**


```python
# Creating NumPy arrays, slicing, matrix multiplication, and vectorization
arr_1d = np.array([10, 20, 30, 40, 50])
arr_2d = np.array([[1, 2, 3], [4, 5, 6]])

print("1D Array:", arr_1d)
print("Shape of 2D Array:", arr_2d.shape)
print("Mean:", np.mean(arr_1d), "| Std Dev:", np.std(arr_1d))

# Matrix multiplication
A = np.random.randn(2, 3)
B = np.random.randn(3, 2)
C = np.dot(A, B)
print("Matrix Product shape (2x3 @ 3x2):", C.shape)

```

### **3. Pandas Essentials: DataFrames & Slicing**


```python
# Creating a sample DataFrame and performing inspection
data = {
    'Student_ID': [101, 102, 103, 104, 105],
    'Age': [20, 21, 19, 22, 20],
    'Study_Hours': [4.5, 6.0, 3.2, 7.5, 5.0],
    'Score': [78, 88, 65, 95, 82]
}

df = pd.DataFrame(data)
print("--- DataFrame Head ---")
print(df.head())
print("\n--- Summary Statistics ---")
print(df.describe())
print("\nFiltered (Study_Hours > 4.5):")
print(df[df['Study_Hours'] > 4.5])

```

### **4. Data Visualization with Matplotlib & Seaborn**


```python
import matplotlib.pyplot as plt

plt.figure(figsize=(12, 5))

# Subplot 1: Matplotlib Scatter Plot
plt.subplot(1, 2, 1)
plt.scatter(df['Study_Hours'], df['Score'], color='crimson', s=100, edgecolors='black')
plt.title("Study Hours vs. Exam Score (Matplotlib)")
plt.xlabel("Study Hours")
plt.ylabel("Exam Score")
plt.grid(True, linestyle='--', alpha=0.6)

# Subplot 2: Seaborn Regression Plot
plt.subplot(1, 2, 2)
sns.regplot(data=df, x='Study_Hours', y='Score', color='teal', marker='o')
plt.title("Linear Trend with 95% CI (Seaborn)")
plt.grid(True, linestyle='--', alpha=0.6)

plt.tight_layout()
plt.show()

```
