### **Aim / Objective:**
To implement techniques for handling missing values, encoding categorical variables, scaling features, and performing feature transformation.

---
### **Theory & Overview:**
Raw data often contains anomalies, missing entries, and disparate scales:
1. **Imputation**: Replacing missing entries using mean, median, mode, or iterative estimators.
2. **Encoding**: Converting categorical variables using One-Hot Encoding (nominal) or Ordinal/Label Encoding (ranked).
3. **Feature Scaling**: Standard Scaler ($z = (x - \mu)/\sigma$) and Min-Max Normalization ($x' = (x - x_{min})/(x_{max} - x_{min})$).
4. **Feature Selection**: Using statistical tests (e.g., ANOVA, Chi-Square) or model-based importance.



### **1. Dataset Creation with Missing Values & Categorical Attributes**


```python
import numpy as np
import pandas as pd

raw_data = {
    'Age': [25, 30, np.nan, 35, 40, np.nan, 28],
    'Salary': [50000, 64000, 58000, np.nan, 85000, 52000, 61000],
    'City': ['Mumbai', 'Delhi', 'Bangalore', 'Delhi', 'Mumbai', 'Bangalore', 'Delhi'],
    'Education': ['Bachelors', 'Masters', 'PhD', 'Bachelors', 'PhD', 'Masters', 'Bachelors'],
    'Purchased': [0, 1, 0, 1, 1, 0, 1]
}

df = pd.DataFrame(raw_data)
print("Raw DataFrame with Missing Values:")
print(df)
print("\nMissing Value Count:")
print(df.isnull().sum())

```

### **2. Handling Missing Data using SimpleImputer**


```python
from sklearn.impute import SimpleImputer

# Numerical Imputation using median
num_cols = ['Age', 'Salary']
imputer = SimpleImputer(strategy='median')
df[num_cols] = imputer.fit_transform(df[num_cols])

print("After Imputing Missing Values with Median:")
print(df)

```

### **3. Encoding Categorical Variables**


```python
from sklearn.preprocessing import OneHotEncoder, OrdinalEncoder

# Ordinal Encoding for ranked education level
edu_order = [['Bachelors', 'Masters', 'PhD']]
ordinal_enc = OrdinalEncoder(categories=edu_order)
df['Education_Encoded'] = ordinal_enc.fit_transform(df[['Education']])

# One-Hot Encoding for nominal City variable
df = pd.get_dummies(df, columns=['City'], drop_first=True, dtype=int)

print("DataFrame after Categorical Encoding:")
print(df)

```

### **4. Feature Scaling (StandardScaler vs MinMaxScaler)**


```python
from sklearn.preprocessing import StandardScaler, MinMaxScaler

scaler_std = StandardScaler()
scaler_minmax = MinMaxScaler()

df_scaled = df.copy()
df_scaled[['Age_Std', 'Salary_Std']] = scaler_std.fit_transform(df[['Age', 'Salary']])
df_scaled[['Age_MinMax', 'Salary_MinMax']] = scaler_minmax.fit_transform(df[['Age', 'Salary']])

print("DataFrame with Scaled Features:")
print(df_scaled[['Age_Std', 'Salary_Std', 'Age_MinMax', 'Salary_MinMax']].head())

```

### **5. Feature Selection using SelectKBest**


```python



from sklearn.feature_selection import SelectKBest, f_classif

X = df.drop(columns=['Education', 'Purchased'])
y = df['Purchased']

selector = SelectKBest(score_func=f_classif, k=3)
X_selected = selector.fit_transform(X, y)

selected_features = X.columns[selector.get_support()]
print("Selected Top 3 Features based on ANOVA F-Score:", list(selected_features))

```
